## Podstawowe instrukcje sterujące

### Selekcja (`if`)

```python
if x < 100:
    print('Wartość jest mniejsza niż 100.')
```

```python
if x > 100:
    print('Wartość jest większa niż 100.')
else:
    print('Wartość nie jest większa niż 100.')
```

```python
if x > 100:
    print('Wartość jest większa niż 100.')
elif x < 50:
    print('Wartość jest mniejsza niż 50.')
else:
    print('Wartość jest z zakresu 50-100.')
```

Blok instrukcji jest wyznaczony przez instrukcje z takim samym wcięciem:

```python
if x < 100:
    print('1. instrukcja')
    print('2. instrukcja')
    if x < 50:
        print('3. instrukcja')
        print('4. instrukcja')
```

### Pętla (`while`)

```python
s = 0
n = 1
while n < 10:
    s += n * n
    n += 1
print(s)
```

### Instrukcja pusta (`pass`)

```python
if x < 100:
    pass
else:
    print('Wartość większa niż 100.')
```

#### Formatowanie napisów (f-strings)
Podstawowe użycie &mdash; jeżeli poprzedzimy literał napisowy znakiem `f`, to wyrażenia występujące wewnątrz `{}` są obliczane, zamieniane na napis i wstawiane w odpowiednie miejsce:

```python
rok, miesiac, dzien = 2000, 10, 14
```

```python
print(f"Data: {rok + 1}/{miesiac}/{dzien}")
```

```python
imie = "Jan"
nazwisko = "Kowalski"
wiek = 30
```

```python
print(f"{imie} {nazwisko} ma {wiek} lat.")
```

Obecnie jest to sugerowany sposób konstruowania napisów z różnych wartości w języku Python. Zazwyczaj daje czytelniejszy kod niż np. sklejanie operatorem `+` lub przekazywanie wielu argumentów do funkcji `print`. Porównaj fragmenty kodu powyżej z równoważnym fragmentami poniżej:

```python
print(str(rok) + '/' + str(miesiac) + '/' + str(rok))
```

```python
print(imie, nazwisko, 'ma', wiek, 'lat.')
```

Zadanie minimalnej liczby znaków jaką ma zajmować pole:

```python
x = 2.34
```

```python
print(f"|{x:10}|")   # liczby domyślnie wyrównane są do prawej
```

```python
napis = "abc"    # napisy domyślnie wyrównane są do lewej
```

```python
print(f"|{napis:5}|")
```

```python
dlugi_napis = "Stoi na stacji lokomotywa"    # Jeżeli napisy nie zmieszczą się w zadanej liczbie znaków,
duza_liczba = 12345678901234567890           # to pola zostaną powiększone.
```

```python
print(f"|{dlugi_napis:3}|{duza_liczba:6}|")
```

```python
print(f"|{x:10}|{napis:5}|{dlugi_napis:3}|{duza_liczba:6}|")
```

Napis ma zajmować co najmniej 8 znaków, ale nie więcej niż 12:

```python
print(f"|{napis:8.12}|{dlugi_napis:8.12}|")
```

Wyrównanie do lewej, do środka, do prawej (niezależnie od typu):

```python
print(f"|{12.32:<10}|{23:^10}|{napis:>10}|")
```

Wypełnienie zadanym znakiem:

```python
print(f"|{12:_<8}|{23:.^8}|{34:0>8}|")
```

Jeżeli parametr formatowania (np. szerokość pola, znak wypełnienia) ma być obliczony (np. na podstawie wartości zmiennej), to należy użyć dodatkowej pary nawiasów klamrowych:

```python
wartosc = 12
znak_wypelnienia = '_'
szerokosc_pola = 10
```

```python
print(f"|{wartosc:{znak_wypelnienia}<{szerokosc_pola}}|")
```

Np. jeżeli potrzebna jest informacja o ciągu kolejnych liczb całkowitych, to nie trzeba przechowywać wszystkich tych liczb w pamięci &mdash; wystarczy informacja o początku i końcu ciągu.

#### Co może być kluczem w słowniku lub elementem zbioru?
Elementy zbioru podobnie jak klucze w słownikach nie muszą być liczbami, ale nie mogą być strukturami danych, które 
mogą zmieniać wartości (np. listami, słownikami). Mogą więc być np. napisami lub krotkami. Krotka użyta jako element zbioru lub klucz w słowniku nie może sama zawierać struktur danych, które mogą zmieniać wartość (np. list).

[Precyzyjniejszy opis tego co może być kluczem lub elementem zbioru](https://docs.python.org/3/glossary.html#term-hashable)

```python
{3, True, "abc", (1, 3, 2), 4.5, (1, 3, 2), 8, True, (4, "abc")}
```

Lista nie może być elementem zbioru:

```python
{3, "abc", [4, 5]}
```

Nie może też być kluczem w słowniku:

```python
{[1, 2]: 'abc'}
```

Krotka, której elementem jest lista nie może należeć do zbioru lub być użyta jako klucz w słowniku:

```python
{(2, []), 4, "abc"}
```

#### (*) Efektywność wyszukiwania elementów w dużych strukturach danych

Wyszukiwanie elementu w zbiorze lub słowniku jest dla dużych struktur danych wielokrotnie szybsze niż wyszukiwanie w liście.

Funkcja do sprawdzenia szybkości wyszukiwania w liście. Dla każdej liczby z zakresu 1&ndash;1000 sprawdza czy należy ona do wcześniej wygenerowanej listy. Zwraca liczbę należących.

```python
def liczba_elementow_z_listy():
    licznik = 0
    for i in range(1000):
        if i in lista:
            licznik += 1
    return licznik
```

```python
liczba_elementow_z_listy()
```

To samo, ale dla zbioru:

```python
def liczba_elementow_ze_zbioru():
    licznik = 0
    for i in range(1000):
        if i in zbior:
            licznik += 1
    return licznik
```

```python
liczba_elementow_ze_zbioru()
```

```python
czas_wyszukania_w_liscie / czas_wyszukania_w_zbiorze
```

Zgłaszanie (wyrzucanie) wyjątków (instrukcja `raise`):

```python
def bmi(kg, cm):
    if kg <= 0:
        raise ValueError("Masa musi być dodatnia.")
    if cm <= 0:
        raise ValueError("Wzrost musi być dodatni.")
    m = cm / 100
    return kg / m ** 2
```

```python
try:
    masa = float(input("Podaj swoją masę w kg: "))
    wzrost = float(input("Podaj swój wzrost w cm: "))
    print(f"BMI: {bmi(masa, wzrost)}")
except ValueError as msg:
    print(msg)
```

### Menedżery kontekstu (instrukcja `with`)

Jeżeli podczas przetwarzania pliku zostanie wyrzucony wyjątek, to plik może pozostać niezamknięty (wyrzucenie wyjątku spowoduje, że przepływ sterowania ominie wywołanie `close`). Żeby temu zapobiec możemy użyć instrukcji <a href="https://docs.python.org/3/tutorial/errors.html#defining-clean-up-actions"><code>finally</code></a>.

Prostszym rozwiązaniem jest użycie instrukcji `with`. Większości operacji zwracających obiekt reprezentujący jakiś zarezerwowany zasób (otwarty plik, otwarte gniazdo sieciowe, połączenie z bazą danych) umożliwia użycie `with` w poniższy sposób (zasób będzie zwolniony automatycznie po zakończeniu działania kodu wewnątrz `with` niezależnie od wyrzuconych i obsłużonych wyjątków):

```python
with open('rzeki.txt', encoding='utf-8') as plik:
    for nazwa in plik:
        print(nazwa, end='')
```

```python
print(plik.closed)
```
